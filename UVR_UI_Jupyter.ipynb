```python
# Asignación de variables
INSTALL_DIR = !pwd
INSTALL_DIR = INSTALL_DIR[0]
MINICONDA_DIR = f"{INSTALL_DIR}/miniconda3"
ENV_DIR = f"{INSTALL_DIR}/env"
MINICONDA_URL = "https://repo.anaconda.com/miniconda/Miniconda3-py313_26.7.1-1-Linux-x86_64.sh"
CONDA_PATH = f"{MINICONDA_DIR}/bin/conda"
```

```python
# Descarga e instalación de Miniconda
import os
from IPython.display import clear_output
if not os.path.exists(CONDA_PATH):
    print("Miniconda not found. Starting download and installation...")
    !wget -O miniconda.sh {MINICONDA_URL}
    !bash miniconda.sh -b -p {MINICONDA_DIR}
    !rm miniconda.sh
    clear_output()
    print("Miniconda installation complete.")
else:
    print("Miniconda already installed. Skipping installation.")
```

```python
# Creación de entorno Conda (Python 3.13, desde conda-forge: no pide aceptar los Terms of Service de Anaconda)
print("Creating Conda environment...")
!{CONDA_PATH} create --no-shortcuts -y --override-channels -c conda-forge --prefix "{ENV_DIR}" python=3.13.16
clear_output()
print("Conda environment created successfully.")
```

```python
# Clonación de UVR5 UI
!git clone https://github.com/Eddycrack864/UVR5-UI.git
clear_output()
print("UVR5 UI cloned successfully.")
```

```python
# Instalación de dependencias (diffq se compila: requiere gcc, p.ej. apt install build-essential)
print("Installing dependencies...")
!{ENV_DIR}/bin/python -m pip install "pip==26.2.1" && \
    {ENV_DIR}/bin/python -m pip install torch==2.7.1 torchvision==0.22.1 --index-url https://download.pytorch.org/whl/cu128 && \
    {ENV_DIR}/bin/python -m pip install -r "{INSTALL_DIR}/UVR5-UI/requirements.txt"
clear_output()
print("Dependencies installation complete.")
```
